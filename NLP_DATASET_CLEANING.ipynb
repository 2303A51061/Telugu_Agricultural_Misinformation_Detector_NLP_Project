```python
# STEP 2: Import required libraries

import pandas as pd
import re
import unicodedata
```

```python
# STEP 3: Read the uploaded Excel file

file_name = list(uploaded.keys())[0]

df = pd.read_excel(file_name)

print("Dataset loaded successfully!")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
# STEP 10: Keep original text

df["original_text"] = df["text"]
```

```python
# STEP 11: Telugu text cleaning function

def clean_text(text):

    # Convert value to string
    text = str(text)

    # Unicode normalization
    text = unicodedata.normalize("NFC", text)

    # Remove URLs
    text = re.sub(r'http\S+|www\S+', '', text)

    # Remove @mentions
    text = re.sub(r'@\w+', '', text)

    # Remove hashtags symbol but keep the word
    text = re.sub(r'#', '', text)

    # Remove emojis and unwanted symbols
    text = re.sub(r'[^\u0C00-\u0C7F0-9A-Za-z\s.,!?%₹]', ' ', text)

    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text)

    # Remove leading and trailing spaces
    text = text.strip()

    return text
```

```python
# STEP 12: Apply cleaning to the text column

df["clean_text"] = df["text"].apply(clean_text)

print("Text cleaning completed!")
```

```python
# STEP 13: Compare original and cleaned text

comparison = df[["original_text", "clean_text"]].head(20)

comparison
```

```python
# STEP 14: Check empty text after cleaning

empty_text = (df["clean_text"].str.strip() == "").sum()

print("Empty text records:", empty_text)
```

```python
# STEP 15: Remove empty text records

before_empty_removal = len(df)

df = df[df["clean_text"].str.strip() != ""]

after_empty_removal = len(df)

print("Records before:", before_empty_removal)
print("Records after:", after_empty_removal)
print("Removed:", before_empty_removal - after_empty_removal)
```

```python
# STEP 16: Remove duplicate cleaned text

before_duplicate_removal = len(df)

df = df.drop_duplicates(subset=["clean_text"])

after_duplicate_removal = len(df)

print("Records before:", before_duplicate_removal)
print("Records after:", after_duplicate_removal)
print("Duplicate records removed:",
      before_duplicate_removal - after_duplicate_removal)
```

```python
# STEP 17: Reset index

df = df.reset_index(drop=True)

df["id"] = range(1, len(df) + 1)

print("ID values updated successfully.")
```

```python
# STEP 18: Check final dataset shape

print("FINAL DATASET")
print("------------------------")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```
